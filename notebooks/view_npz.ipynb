To visualize npz file

Data Loader

```python
import json
import numpy as np
from pathlib import Path
```

```python

repo_root = Path.cwd().parent

processed_root = repo_root / "data" / "ecg" / "processed" / "v2" 

t = "train"

label_data = np.load(processed_root / t / "labels" / "label_shard_000.npz")
td_data = np.load(processed_root / t / "time_domain_data" / "td_shard_000.npz")
fd_data = np.load(processed_root / t / "frequency_domain_data" / "fd_shard_000.npz")

frequency_bins = np.load(processed_root / "frequency_bins.npz")["data"]

with open(processed_root / "label_map.json", "r") as f:
    label_map = json.load(f)
```

Print Keys

```python
print("label keys:", label_data.files)
print("td keys:", td_data.files)
print("fd keys:", fd_data.files)
```

Inspect shapes and dtypes

```python
for name, data in [("label", label_data), ("td", td_data), ("fd", fd_data)]:
    print(f"\n{name.upper()}")
    for key in data.files:
        arr = data[key]
        print(f"{key}: shape={arr.shape}, dtype={arr.dtype}")
```

To verify the stored content

```python
label_arr = label_data["data"]
td_arr = td_data["data"]
fd_arr = fd_data["data"]

print("label_map:", label_map)
print("frequency_bins shape:", frequency_bins.shape)

print("\nFirst 10 labels:")
print(label_arr[:10])

print("\nOne TD sample shape:", td_arr[0].shape)
print("One FD sample shape:", fd_arr[0].shape)

print("\nTD sample 0, channel 0, first 10 values:")
print(td_arr[0, 0, :10])

print("\nFD sample 0, channel 0, first 5 frequency bins [magnitude, phase]:")
print(fd_arr[0, 0, :5, :])
```

Frequency Bin Inspection

```python
print("Frequency bins shape:", frequency_bins.shape)
print("Frequency resolution (Hz):", frequency_bins[1] - frequency_bins[0])

print("\nFirst 10 frequency bins (Hz):")
print(frequency_bins[:10])

print("\nLast 10 frequency bins (Hz):")
print(frequency_bins[-10:])

print("\nMin / Max frequency (Hz):")
print(frequency_bins[0], frequency_bins[-1])
```

Label Map Inspection

```python
unique_labels, counts = np.unique(label_arr, return_counts=True)

inv_label_map = {v: k for k, v in label_map.items()}

print("Label distribution:")
for label_id, count in zip(unique_labels, counts):
    print(f"{label_id:2d}  {inv_label_map[label_id]:>2s}  {count}")
```
